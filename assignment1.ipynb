```python
#In C and Java, variables have a fixed data type.
#If x is declared as an int, assigning a floating-point value like 4.5 causes a type mismatch,
#so the compiler reports an error.

#In Python, variables are dynamically typed, meaning a variable can refer to objects of different types at different times.
# Therefore, assigning 3 and then 4.5 to the same variable is allowed.
```
