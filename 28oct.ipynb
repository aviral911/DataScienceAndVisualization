```python
import numpy as np
import os
import pandas as pd


values = np.random.randint(1, 20, size=20, dtype=np.int32)

values.tofile("numbers.bin")

print("Values:", values)
print("File size:", os.path.getsize("numbers.bin"), "bytes")
```

```python
import numpy as np
import os

arr = np.random.randint(1, 100, size=(5, 5))

np.save("array.npy", arr)

loaded_arr = np.load("array.npy")

print("Original array:")
print(arr)

print("\nLoaded array:")
print(loaded_arr)

print("File size:", os.path.getsize("array.npy"), "bytes")
```
