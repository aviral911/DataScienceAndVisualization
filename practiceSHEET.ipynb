```python
import pandas as pd
from datetime import datetime
```

```python
#1
dt = datetime(2026, 2, 15, 10, 30)
print("Year:", dt.year, "| Month:", dt.month, "| Day:", dt.day,
      "| Hour:", dt.hour, "| Minute:", dt.minute)
```

```python
#2
dt1 = datetime(2026, 2, 15, 10, 0)
dt2 = datetime(2026, 2, 15, 12, 30)

difference = dt2 - dt1

print("Difference:", difference)
print("Hours:", difference.seconds // 3600)
print("Minutes:", (difference.seconds % 3600) // 60)
```

```python
#q3
date_string = "20/02/2026"

dt = datetime.strptime(date_string, "%d/%m/%Y")

result = dt.strftime("%Y-%m-%d")

print(result)
```

```python

dates = ["2026-02-01", "2026-02-05", "2026-02-10"]

result = pd.to_datetime(dates)

print(result)
```

```python

dates = ["2026-02-01", "invalid", "2026-02-05", None]

result = pd.to_datetime(dates, errors="coerce")

print(result)
print("Missing values:")
print(result.isna())
```

```python

dates = pd.to_datetime([
    "2026-02-01",
    "2026-02-02",
    "2026-02-04",
    "2026-02-06",
    "2026-02-08"
])

sales = pd.Series(
    [100, 150, 120, 180, 200],
    index=dates,
    name="units_sold"
)
```

```python
print(sales.index)
print(type(sales.index))
```

```python
print(sales.loc["2026-02-04"])
```

```python
print("First observation:")
print(sales.iloc[0])

print("Last observation:")
print(sales.iloc[-1])
```

```python
result = sales.loc["2026-02-02":"2026-02-06"]

print(result)
```

```python
print("Total sales:", sales.sum())
print("Average sales:", sales.mean())
```

```python

dates = pd.date_range(
    "2026-02-01",
    periods=6,
    freq="D"
)

daily_sales = pd.Series(
    [100, 120, 150, 130, 180, 200],
    index=dates
)

print(daily_sales)
```

```python
df = pd.DataFrame({
    "Current Sales": daily_sales,
    "Previous Sales": daily_sales.shift(1)
})

print(df)
```

```python
df = pd.DataFrame({
    "Current Sales": daily_sales,
    "Next Sales": daily_sales.shift(-1)
})

print(df)
```

```python
change = daily_sales - daily_sales.shift(1)

print(change)
```

```python
shifted = daily_sales.shift(2, freq="D")

print("Original Series:")
print(daily_sales)

print("\nShifted Series:")
print(shifted)
```
