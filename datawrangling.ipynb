```python
import pandas as pd
import numpy as np
import matplotlib as mp
```

```python
# Create a dictionary
data = {
    "department" : ["CSE" , "IT" , "DS" ,  "ML","AI"],
    "YEAR":[1,4,2,4,3],
     "Enrolled_Students": [120, 110, 80, 90,221]

}# Convert dictionary into DataFrame

df=pd.DataFrame(data)
# Set Department and Year as hierarchical index
result = df.set_index(["department","YEAR"])
print(result)
# Select all records of CSE department
print(result.loc["CSE"])
# Select the average marks for CSE, Semester 2
print(result.loc[("CSE", 1), "Enrolled_Students"])
```

```python
student = pd.DataFrame({
    "ID":[101,102,103,104,105],
    "name":["Aviral","suhas","kartik","naman","abhinav"]
})
#marks df
marks=pd.DataFrame({
    "ID":[101,102,103,104,105],
    "marks":[85,96,92,90,23]

})
#innner join
inner = pd.merge(student,marks,on="ID",how="inner")
print("inner join: ")
print(inner)
 # Perform Outer Join (all records from both DataFrames)
outer = pd.merge(student, marks,on="ID",how="outer")
print("outer join: ")
print(outer)
# Keep only IDs present in both datasets
print(" Keep only IDs present in both datasets")
data1= pd.merge(student,marks,on="ID",how="inner")
print(data1)
# Keep every student; unmatched marks become NaN
print("Keep every student; unmatched marks become NaN")
left = pd.merge(student, marks,on="ID",how="left")
print(left)
# Keep all IDs from both datasets
print("Keep all IDs from both datasets")

outer = pd.merge(student , marks,on= "ID",how="outer")
print(outer)
```

```python
#3.2 concat(): Stack datasets
jan = pd.DataFrame({
    "month":["jan","jan"],
    "product":["pen", "book"],
    "units":[120,190]
})
feb = pd.DataFrame({
 "Month": ["Feb", "Feb"],
 "Product": ["Pen", "Notebook"],
 "Units": [150, 95]
})
sales = pd.concat([jan,feb],ignore_index=True)
print (sales)
```

```python
#3.3 join(): Combine using index labels
students_info= student.set_index("ID")
student_marks = marks.set_index("ID")
joined=students_info.join(student_marks,how="left")
print(joined)
```

```python
#4.1 pivot(): Long format to wide format
sales_long = pd.DataFrame({
 "Product": ["Pen", "Pen", "Notebook", "Notebook"],
 "Month": ["Jan", "Feb", "Jan", "Feb"],
 "Units_Sold": [120, 150, 80, 95]
})
sales_wide=sales_long.pivot(index="Product",columns="Month",values="Units_Sold")
print(sales_wide)
```
