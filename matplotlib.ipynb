```python
import  matplotlib.pyplot as plot
```

```python
import numpy as np
```

```python
import pandas as pd
```

```python
plot.figure(figsize=(8, 4))
plot.plot([1, 2, 3], [4, 6, 5])
plot.show()
```

```python
x=[1,2,3,4,5,6]
y = [10,20,23,45,57,60]
plot.plot(x,y,marker='o')
plot.title('simple line plot')
plot.xlabel('X')
plot.grid(alpha=0.25)
plot.show()
plot.title('Monthly Sales')
plot.xlabel('Month')
plot.ylabel('Sales (Rs)')

```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
online = [20, 25, 30, 28, 35, 40]
offline = [15, 18, 22, 25, 24, 30]

plot.plot(months, online, marker='o', linestyle='-', linewidth=2, label='Online')
plot.plot(months, offline, marker='s', linestyle='--', linewidth=2, label='Offline')

plot.title('Monthly Sales')
plot.xlabel('Month')
plot.ylabel('Sales (Rs)')
plot.legend()
plot.grid(axis='y', linestyle='--', alpha=0.25)
plot.xticks(rotation=30)
plot.show()
```

```python
x = list(range(1, 13))
y = [40, 54, 34, 22, 65, 445, 54, 23, 55, 23, 55, 1]

plot.plot(x, y, marker='o')

plot.xlim(0, 12)
plot.ylim(0, 100)

plot.grid(axis='y', linestyle='--', alpha=0.25)

plot.annotate(
    'peak',
    xy=(6, 65),
    xytext=(8, 90),
    arrowprops={'arrowstyle': '->'}
)

plot.show()
```

```python
categories = ['Beverages', 'Snacks', 'Meals', 'Stationery']
values = [82000, 61000, 97000, 54000]
plot.bar(categories, values)
plot.ylabel('Sales (Rs)')
plot.xticks(rotation=20)
plot.show()
```

```python
plot.barh(categories, values)
plot.xlabel('Sales (Rs)')
plot.show()
```

```python
categories = ['Beverages', 'Snacks', 'Meals', 'Stationery']
sem1 = [21, 16, 25, 13]
sem2 = [25, 19, 30, 17]

x = np.arange(len(categories))
width = 0.35

plot.bar(x - width/2, sem1, width, label='Sem 1')
plot.bar(x + width/2, sem2, width, label='Sem 2')

plot.title('Grouped bar chart')
plot.xticks(x, categories, rotation=30)
plot.legend()
plot.show()
```

```python
labels = ['Jan', 'Feb', 'Mar']
a = [10, 12, 14]   # UPI
b = [5, 6, 7]      # Card
c = [3, 4, 5]      # Cash

plot.bar(labels, a, label='UPI')
plot.bar(labels, b, bottom=a, label='Card')
plot.bar(labels, c, bottom=np.array(a) + np.array(b), label='Cash')

plot.title('Stacked bar: payment composition')
plot.legend()
plot.show()
```

```python
labels = ['Group A', 'Group B', 'Group C']

a = np.array([30, 20, 50])
b = np.array([50, 40, 30])
c = np.array([20, 40, 20])

total = a + b + c

a_pct = a / total * 100
b_pct = b / total * 100
c_pct = c / total * 100

plot.bar(labels, a_pct, label='Category A')
plot.bar(labels, b_pct, bottom=a_pct, label='Category B')
plot.bar(labels, c_pct, bottom=a_pct + b_pct, label='Category C')

plot.title('100% stacked bar')
plot.ylabel('Percentage (%)')
plot.legend()
plot.show()
```

```python
labels = ['UPI', 'Card', 'Cash', 'Wallet']
sizes = [48, 32, 14, 6]

plot.pie(sizes, labels=labels, autopct='%1.0f%%', startangle=90)
plot.title('Payment share')
plot.show()
```

```python
orders = [42, 45, 48, 50, 53, 56, 60, 63, 67, 71, 74, 79, 84]
sales = [3100, 3400, 3550, 3900, 4200, 4500, 4700, 5100, 5600, 5900, 6300, 6700, 6900]

plot.scatter(orders, sales)
plot.title('Orders vs sales')
plot.xlabel('Daily orders')
plot.ylabel('Daily sales (Rs)')
plot.show()
```

```python
orders = [42, 45, 48, 50, 53, 56, 60, 63, 67, 71, 74, 79, 84]
sales = [3100, 3400, 3550, 3900, 4200, 4500, 4700, 5100, 5600, 5900, 6300, 6700, 6900]
bubble_size = [50, 80, 100, 120, 150, 180, 200, 220, 250, 280, 300, 330, 350]

plot.scatter(orders, sales, s=bubble_size, alpha=0.6)
plot.title('Bubble-style scatter plot')
plot.xlabel('Daily orders')
plot.ylabel('Daily sales (Rs)')
plot.show()
```

```python
order_values = [120, 150, 180, 200, 210, 230, 250, 260, 270, 280,
                300, 310, 320, 330, 340, 350, 360, 370, 380, 400,
                420, 450, 480, 520, 560, 600, 650, 700, 850]

plot.hist(order_values, bins=10)
plot.title('Distribution of order values')
plot.xlabel('Order value')
plot.ylabel('Frequency')
plot.show()
```

```python
plot.hist(order_values, bins=10, density=True)
plot.title('Density histogram')
plot.xlabel('Order value')
plot.ylabel('Density')
plot.show()
```

```python
beverages = [60, 70, 75, 80, 85, 90, 95, 100, 110, 120, 180, 220]
meals = [150, 180, 200, 220, 240, 250, 260, 280, 300, 320, 350, 370, 420]
stationery = [40, 50, 55, 60, 65, 70, 75, 80, 90, 100, 150, 180]

data = [beverages, meals, stationery]

plot.boxplot(data, labels=['Beverages', 'Meals', 'Stationery'])
plot.title('Comparing distributions with box plots')
plot.ylabel('Order value')
plot.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 14, 12, 18, 21]

fig, ax = plot.subplots(figsize=(8, 4))

ax.plot(x, y, marker='o')
ax.set_title('Simple Line Plot')
ax.set_xlabel('X')
ax.set_ylabel('Value')
ax.grid(alpha=0.25)

plot.show()
```
